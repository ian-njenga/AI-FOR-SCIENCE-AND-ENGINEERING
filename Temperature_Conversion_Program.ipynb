```python
#The program is created for a user to check their temperature
#Created a while loop that allows the user to input more than one value
try:
  while True:
    #Asks the user to input the value of their temperature
    temp = float(input('Enter your temperature value in °C to convert to °F:'));

    def converted(temp):
      #Converts the temperature of the user from °C to °F
      return temp * 1.8 + 32
    #Prints the temperature value of the user
    converted_temp = converted(temp);
    print('Your temperature is: ', converted_temp, '°F');
    #Checks the temperature of the user against a range of 97°F and 99°F
    if (converted_temp<=97):
      print('Your temperature is low, kindly visit a doctor!');
    elif (converted_temp>=99):
      print('You might have a fever, kindly visit a doctor!');
    else:
      print('You seem to be healthy');
except KeyboardInterrupt:
    print('Hope you are well.\nProgram terminated')
```
